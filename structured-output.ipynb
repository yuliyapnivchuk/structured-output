```python
import os
from langchain_groq import ChatGroq

llm = ChatGroq(model_name="mixtral-8x7b-32768")
```

```python
from pydantic import BaseModel
from langchain.output_parsers import PydanticOutputParser
from langchain.prompts import ChatPromptTemplate

class NutritionInfo(BaseModel):
    food: str
    benefits: str
    vitamins: list[str]

parser = PydanticOutputParser(pydantic_object=NutritionInfo)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a nutritionist. Answer in JSON format."),
    ("human", "Tell me about {food} in a structured format.")
])

prompt_with_parser = prompt.format_prompt(food="spinach").to_string() + f"\n{parser.get_format_instructions()}"
```

```python
response = llm.invoke(prompt_with_parser)

structured_response = parser.parse(response.content)

print(structured_response)
```
